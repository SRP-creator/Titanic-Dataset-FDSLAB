```python
sex = df['Sex']
name = df['Name']

match = 0
total = 0

for s, n in zip(sex, name):

    if pd.isna(s) or pd.isna(n):
        continue

    total += 1

    if s == n:
        match += 1

similarity = match / total
dissimilarity = 1 - similarity

print("\nNominal Attribute Similarity")
print("Sex vs Name Similarity =", similarity)

print("\nNominal Attribute Dissimilarity")
print("Sex vs Name Dissimilarity =", dissimilarity)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity

# Select features
X = df[['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].fillna(df['Age'].median())

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Compute similarity matrix
similarity_matrix = cosine_similarity(X_scaled)

print(similarity_matrix.shape) 
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

similarity_df = pd.DataFrame(similarity_matrix[:20, :20])

plt.figure(figsize=(10, 8))
sns.heatmap(similarity_df, cmap='coolwarm')
plt.title("Similarity Matrix (First 20 Passengers)")
plt.show()
```
